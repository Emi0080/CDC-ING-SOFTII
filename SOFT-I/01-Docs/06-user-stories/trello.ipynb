```python
!pip install openpyxl
import json
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import Alignment

# Leer el JSON
with open("YxDG1h4x - hufase2-cdc (3).json", "r", encoding="utf-8") as f:
    board = json.load(f)

# Diccionario de listas
lists = {l["id"]: l["name"] for l in board["lists"]}

# Diccionario de etiquetas
labels = {l["id"]: l["name"] for l in board["labels"]}

# Diccionario de checklists "Criterios de Aprobación"
criterios = {}

for checklist in board["checklists"]:
    if checklist["name"].strip().lower() == "criterios de aceptación":
        criterios[checklist["idCard"]] = [
            item["name"] for item in checklist["checkItems"]
        ]

# Crear registros
datos = []

for card in board["cards"]:

    datos.append({
        "Nombre": card["name"],
        "Etiquetas": ", ".join(labels[i] for i in card.get("idLabels", [])),
        "Descripcion": card.get("desc", ""),
        "Criterios de Aprobacion": "\n".join(
            f"• {item}" for item in criterios.get(card["id"], [])
        ),
        "Last Activity Date": card.get("dateLastActivity"),
        "List Name": lists.get(card["idList"], "")
    })

# Crear DataFrame
df = pd.DataFrame(datos)


# Guardar a Excel
archivo = "HistoriasUsuario.xlsx"
df.to_excel(archivo, index=False)

# Activar ajuste de texto (Wrap Text)
wb = load_workbook(archivo)
ws = wb.active

for row in ws.iter_rows():
    for cell in row:
        cell.alignment = Alignment(wrap_text=True, vertical="top")

# (Opcional) Ajustar ancho de columnas
for col in ws.columns:
    max_length = 0
    column = col[0].column_letter

    for cell in col:
        try:
            if len(str(cell.value)) > max_length:
                max_length = len(str(cell.value))
        except:
            pass

    ws.column_dimensions[column].width = min(max_length + 2, 60)

wb.save(archivo)

df.head(1000)
```
